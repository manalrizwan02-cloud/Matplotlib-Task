```python
#task 5

#concepts:
#Line Plot
#Adding Legend

#Method(marker="o")

subjects = ["Math", "Physics", "Computer", "English"]

Ali = [78, 82, 90, 75]
Sara = [88, 76, 95, 85]

plt.plot(subjects,Ali, marker="o", label="Ali")
plt.plot(subjects,Sara, marker="o", label="Sara")

plt.title("Ali vs Sara Performance")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.legend()

plt.show()
```

```python
#Task 6
#concepts:
#Line Plot
#Adding Legend
#Adding Grid Line
#Method(marker="s")

Months = ["Jan", "Feb", "Mar", "Apr", "May"]

Karachi = [100, 130, 120, 160, 180]
Lahore = [90, 110, 140, 150, 170]
Islamabad = [80, 100, 125, 145, 160]

plt.plot(Months,Karachi,marker="s",label="Karachi")
plt.plot(Months,Lahore,marker="s",label="Lahore")
plt.plot(Months,Islamabad,marker="s",label="Islamabad")

plt.title("Shop Branch sales Comparison")

plt.xlabel("Months")
plt.ylabel("Branch")

plt.legend()

plt.grid()
plt.show()
```

```python
#Task 7
#Concept:
#Bar plot
#Adding Legend
#Using numpy for width calculate
#Using width for overlap bar plot
#using xtick for show product  position

import numpy as np
import matplotlib.pyplot as plt


products = ["Laptop", "Mobile", "Tablet", "Headphones"]

Online = [40, 70, 35, 80]
In_Store = [30, 50, 45, 60]

x = np.arange(len(products))
width = 0.4

plt.bar(x - width/2, Online, width, label="Online")
plt.bar(x + width/2, In_Store, width, label="InStore")

plt.xticks(x,products)
plt.title("Products by Online vs Instore Comparison")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

Departments = ["IT", "HR", "Marketing", "Finance"]

January = [40, 30, 35, 25]
February = [50, 35, 45, 30]
March = [60, 40, 55, 35]

x = np.arange(len(Departments))
width = 0.25

plt.figure(figsize=(8, 5))

plt.bar(x - width, January, width,label="January")
plt.bar(x, February, width, label="February")
plt.bar(x + width, March, width, label="March")

plt.xticks(x, Departments)
plt.title("Department Comparison")
plt.xlabel("Departments")
plt.ylabel("Students")
plt.legend()

plt.show()
```
